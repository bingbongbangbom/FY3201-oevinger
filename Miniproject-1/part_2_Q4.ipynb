```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "data/Land_and_Ocean_complete.txt",
    comment="%",
    sep=r"\s+",
    skiprows=2,
    header=None
)

df.columns = [
    "Year", "Month",
    "Monthly_Anomaly", "Monthly_Unc",
    "Annual_Anomaly", "Annual_Unc",
    "FiveYear_Anomaly", "FiveYear_Unc",
    "TenYear_Anomaly", "TenYear_Unc",
    "TwentyYear_Anomaly", "TwentyYear_Unc"
]

# Calculate average monthly anomaly for each year
yearly_avg = df.groupby("Year")["Monthly_Anomaly"].mean()

# Convert to NumPy array
arr = yearly_avg.reset_index().to_numpy()



arr = arr[-45:]
print(arr)
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    arr_fat[:, 1],
    arr[:, 1],
    label="observed annual data",
    color="blue"
)

slope, intercept, r_value, p_value, std_err = sp.stats.linregress(
    arr_fat[:, 1], arr[:, 1]
)

plt.plot(
    arr_fat[:, 1],
    slope * arr_fat[:, 1] + intercept,
    color="red",
    label="Linear Regression"
)
r = np.corrcoef(arr_fat[:, 1], arr[:, 1])[0, 1]

print(f"Pearson correlation: {r:.4f}")
print(f"R²: {r**2:.4f}")
print(f"p-value: {p_value:.4g}")
# Statistical values
text = (
    f"Pearson correlation: {r:.4f}\n"
    f"Slope = {slope:.4f}\n"
    f"Intercept = {intercept:.4f}\n"
    f"R² = {r_value**2:.4f}\n"
    f"p-value = {p_value:.4g}\n"
    f"Std. error = {std_err:.4f}"
)

# Add statistical box
plt.text(
    0.05, 0.95,
    text,
    transform=plt.gca().transAxes,
    verticalalignment="top",
    bbox=dict(
        boxstyle="round",
        facecolor="white",
        alpha=0.8
    )
)

plt.grid()
plt.xlabel("Obesity Rate [%]")
plt.ylabel(r"Temperature Anomaly [$^\circ$C]")
plt.title("Relationship between Obesity in Afghanistan and Temperature")
plt.legend()

plt.show()
```
